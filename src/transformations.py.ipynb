```python

from pyspark.sql import DataFrame
from pyspark.sql.functions import col, trim, element_at, split, concat_ws

def clean_patients(df: DataFrame) -> DataFrame:
    """Standardizes patient names and extracts state from birthplace."""
    return (
        df.filter(col("patient").isNotNull())
        .dropDuplicates(["patient"])
        .withColumn("patient_name", trim(concat_ws(" ", col("first"), col("last"))))
        .withColumn("state", trim(element_at(split(col("birthplace"), r"\s+"), -2)))
    )

def filter_invalid_claims(df: DataFrame) -> DataFrame:
    """Filters out claims with missing IDs or negative costs."""
    return df.filter(col("claim_id").isNotNull() & (col("total_cost") >= 0))
```
